```python
import json
import re
from collections import defaultdict
```

```python
def extract_experiments_count(filename):
    """Извлекает количество экспериментов из имени файла."""
    match = re.search(r'_(\d+)exp', filename)
    return int(match.group(1)) if match else 1
```

```python
def make_data_for_special_codes(filename):
    with open(filename, 'r') as f:
        data = f.read()
        obj = json.loads(data)

        dims_freqs = defaultdict(lambda: defaultdict(lambda: defaultdict(float)))
        for key, values in obj.items():

            for delta_range, exps in values:
                if delta_range[1] != 0:
                    continue

                for exp in exps:
                    dims_freqs[key][delta_range[0]][exp[2]] += 1

                experiments_count = len(exps)
                for dim in dims_freqs[key][delta_range[0]]:
                    dims_freqs[key][delta_range[0]][dim] = dims_freqs[key][delta_range[0]][dim] / experiments_count

    new_filename = filename.replace('.json', '_special.txt')
    with open(new_filename, 'w') as f:
        for key in dims_freqs:
            f.write(f"{key}:\n")

            for l_range in dims_freqs[key]:
                f.write(f"  {l_range}:\n")
                for dim in dims_freqs[key][l_range]:
                    f.write(f"    {dim}: {dims_freqs[key][l_range][dim]}\n")
```

```python
make_data_for_special_codes('./RM(2_7)_L10_R14_E100/hulls.json')
```
